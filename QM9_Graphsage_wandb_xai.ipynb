```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
dataset = QM9(root ="data/qm9")
```

```python
print(f"  Number of graphs: {len(dataset)}")
print(f"  Number of node features: {dataset.num_node_features}")
print(f"  Number of edge features: {dataset.num_edge_features}")
print(f"  Number of targets: {dataset.num_classes}")
print(f"  Number of features: {dataset.num_features}")

sample = dataset[0]
print(f"\nExample molecule:")
print(f"  Name       : {sample.name}")
print(f"  Atoms      : {sample.x.shape[0]}")
print(f"  x shape    : {sample.x.shape}   (atoms x features)")
print(f"  edge_index : {sample.edge_index.shape}")
print(f"  edge_attr  : {sample.edge_attr.shape}")
print(f"  pos        : {sample.pos.shape}")
print(f"  y          : {sample.y}")
```

```python
from torch.utils.data import random_split

num_graphs = len(dataset)
train_size = int(0.8 * num_graphs)
validation_size = int(0.1 * num_graphs)
test_size = num_graphs - train_size - validation_size

torch.manual_seed(42)

train_dataset, validation_dataset, test_dataset = random_split(
    dataset,
    [train_size, validation_size, test_size]
)

HIDDEN_DIM  = 64
```

```python
NUM_TARGETS = 19

TARGET_NAMES = [
    "mu", "alpha", "HOMO", "LUMO", "gap", "R2", "ZPVE",
    "U0", "U", "H", "G", "Cv",
    "U0_atom", "U_atom", "H_atom", "G_atom",
    "A", "B", "C"
]

TARGET_UNITS = [
    "D", "a0^3", "eV", "eV", "eV", "a0^2", "eV",
    "eV", "eV", "eV", "eV", "cal/mol/K",
    "eV", "eV", "eV", "eV",
    "GHz", "GHz", "GHz"
]
```

```python
all_train_y = torch.cat([data.y for data in train_dataset], dim=0)

target_mean = all_train_y.mean(dim=0)
target_std  = all_train_y.std(dim=0).clamp(min=1e-6)

print("Per-target statistics (train set):")
print(f"{'idx':>4} {'name':>10} {'mean':>12} {'std':>12} {'unit'}")
print("-" * 55)
for i, (name, unit) in enumerate(zip(TARGET_NAMES, TARGET_UNITS)):
    print(f"{i:>4} {name:>10} {target_mean[i].item():>12.4f} {target_std[i].item():>12.4f}  {unit}")
```

```python
def normalize_dataset(subset):
    normalized = []
    for data in subset:
        data.y = (data.y - target_mean) / target_std
        normalized.append(data)
    return normalized

train_dataset      = normalize_dataset(train_dataset)
validation_dataset = normalize_dataset(validation_dataset)
test_dataset       = normalize_dataset(test_dataset)
```

```python
batch_size = 64
train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=batch_size,
    shuffle=False)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False)

print(f"Train batches : {len(train_loader)}")
print(f"Validation   batches : {len(validation_loader)}")
print(f"Test  batches : {len(test_loader)}")
```

```python
class GraphSAGE(torch.nn.Module):
    def __init__(self):
        super().__init__()

        # Input embedding
        self.input_lin = Linear(dataset.num_node_features, HIDDEN_DIM)

        # GraphSAGE layers
        self.conv1 = SAGEConv(HIDDEN_DIM, HIDDEN_DIM)
        self.bn1   = BatchNorm1d(HIDDEN_DIM)
        self.conv2 = SAGEConv(HIDDEN_DIM, HIDDEN_DIM)
        self.bn2   = BatchNorm1d(HIDDEN_DIM)

        self.lin1 = Linear(HIDDEN_DIM, HIDDEN_DIM // 2)
        self.lin2 = Linear(HIDDEN_DIM // 2, NUM_TARGETS)

    def forward(self, x, edge_index, batch):
        x = self.input_lin(x).relu()

        x = self.conv1(x, edge_index)
        x = self.bn1(x).relu()

        x = self.conv2(x, edge_index)
        x = self.bn2(x).relu()

        x = global_mean_pool(x, batch)

        x = self.lin1(x).relu()
        x = self.lin2(x)
        return x

model = GraphSAGE().to(device)

print(model)
print(f"\nTrainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.001)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.75,
    patience=10,
    min_lr=1e-6
)

_mean = target_mean.to(device)
_std  = target_std.to(device)


def train(loader):
    model.train()
    total_loss = 0
    for batch in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        out  = model(batch.x, batch.edge_index, batch.batch)
        loss = F.mse_loss(out, batch.y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        total_loss += loss.item() * batch.num_graphs
    return total_loss / len(loader.dataset)


@torch.no_grad()
def evaluate(loader):
    model.eval()
    total_mae = torch.zeros(NUM_TARGETS)

    for batch in loader:
        batch = batch.to(device)
        out   = model(batch.x, batch.edge_index, batch.batch)

        pred = out  * _std + _mean
        true = batch.y * _std + _mean

        total_mae += (pred - true).abs().sum(dim=0).cpu()

    mae_per_target = total_mae / len(loader.dataset)
    mean_mae       = mae_per_target.mean().item()
    return mae_per_target, mean_mae
```

```python
wandb.init(
    project="qm9-graphsage",
    config={
        "model": "GraphSAGE",
        "hidden_dim": 64,
        "num_layers": 2,
        "lr": 0.001,
        "batch_size": 64,
        "epochs": 100,
        "optimizer": "Adam",
        "num_node_features": dataset.num_node_features,
        "trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
        "normalization": "per_target_zscore",
        "pooling": "global_mean_pool",
    }
)

wandb.watch(model, log="all", log_freq=10)
```

```python
best_val_mae = float('inf')
best_test_mae_per_target = None

history = {'train_loss': [], 'val_mae': [], 'test_mae': [], 'lr': []}

for epoch in range(1, 101):
    train_loss              = train(train_loader)
    val_mae_per_t, val_mae  = evaluate(validation_loader)
    test_mae_per_t, test_mae = evaluate(test_loader)

    scheduler.step(val_mae)

    current_lr = optimizer.param_groups[0]['lr']

    history['train_loss'].append(train_loss)
    history['val_mae'].append(val_mae)
    history['test_mae'].append(test_mae)
    history['lr'].append(current_lr)

    log_dict = {
            "epoch": epoch,
            "train_loss": train_loss,
            "val_mae": val_mae,
            "test_mae": test_mae,
            "lr": current_lr
        }
    for i, name in enumerate(TARGET_NAMES):
        log_dict[f"val_mae/{name}"]  = val_mae_per_t[i].item()
        log_dict[f"test_mae/{name}"] = test_mae_per_t[i].item()

    wandb.log(log_dict)

    if val_mae < best_val_mae:
        best_val_mae             = val_mae
        best_test_mae_per_target = test_mae_per_t.clone()
        torch.save(model.state_dict(), 'best_model.pt')
        wandb.save('best_model.pt')

    if epoch % 10 == 0:
        print(f"Epoch {epoch:>3} | Loss: {train_loss:.4f} | "
              f"Val MAE: {val_mae:.4f} | Test MAE: {test_mae:.4f} | LR: {current_lr:.2e}")

print(f"\nBest mean Val MAE  : {best_val_mae:.4f}")
```

```python
class SingleTargetWrapper(torch.nn.Module):
    def __init__(self, model, target_idx):
        super().__init__()
        self.model = model
        self.target_idx = target_idx

    def forward(self, x, edge_index, batch):
        out = self.model(x, edge_index, batch)
        return out[:, self.target_idx].unsqueeze(1)
```

```python
import networkx as nx
from torch_geometric.utils import to_networkx
from collections import defaultdict

model.load_state_dict(torch.load("best_model.pt"))
model.eval()

all_explanations = {}
edge_masks_all = []
node_masks_all = []

for target_idx, target_name in enumerate(TARGET_NAMES):
    wrapper = SingleTargetWrapper(model, target_idx).to(device)

    explainer = Explainer(
        model=wrapper,
        algorithm=GNNExplainer(epochs=200),
        explanation_type="model",
        node_mask_type="attributes",
        edge_mask_type="object",
        model_config=dict(
            mode="regression",
            task_level="graph",
            return_type="raw",
        ),
    )

    explanation = explainer(
        x=sample.x,
        edge_index=sample.edge_index,
        batch=torch.zeros(sample.x.size(0), dtype=torch.long, device=device),
    )

    all_explanations[target_name] = explanation
    edge_masks_all.append(explanation.edge_mask.detach().cpu().numpy())
    node_masks_all.append(explanation.node_mask.detach().cpu().numpy())
```

```python
def aggregate_undirected_edges(edge_index, edge_mask):
    edge_groups = defaultdict(list)
    edge_index_np = edge_index.cpu().numpy()
    for idx in range(edge_index_np.shape[1]):
        u, v = int(edge_index_np[0, idx]), int(edge_index_np[1, idx])
       edge_groups[tuple(sorted((u, v)))].append(edge_mask[idx])

    undir_edges = list(edge_groups.keys())
    undir_weights = np.array([np.mean(v) for v in edge_groups.values()])
    return undir_edges, undir_weights


aggregated_edge_masks = []
for edge_mask in edge_masks_all:
    undir_edges, undir_weights = aggregate_undirected_edges(sample.edge_index, edge_mask)
    aggregated_edge_masks.append(undir_weights)
aggregated_edge_masks = np.stack(aggregated_edge_masks)

node_masks_all = np.stack(node_masks_all)
node_feature_matrix = node_masks_all.mean(axis=1)

FEATURE_NAMES = FEATURE_NAMES = [
    "atom_type_H", "atom_type_C", "atom_type_N", "atom_type_O",
    "atom_type_F", "atomic_num", "aromaticity", "hybrid_SP",
    "hybrid_SP2", "hybrid_SP3", "num_Hs"
    ]

G = nx.Graph()

for u, v in undir_edges:
    G.add_edge(u, v)
pos = nx.spring_layout(G,seed=42)


fig1, ax1 = plt.subplots(figsize=(14, 6))
im1 = ax1.imshow(aggregated_edge_masks, aspect="auto", cmap="YlOrRd")
ax1.set_yticks(range(len(TARGET_NAMES)))
ax1.set_yticklabels([f"{n} ({u})" for n, u in zip(TARGET_NAMES, TARGET_UNITS)], fontsize=9)
ax1.set_xlabel("Undirected edge index")
ax1.set_title("Edge importance heatmap\n(GNNExplainer + GraphSAGE)")
plt.colorbar(im1, ax=ax1, label="importance")
plt.tight_layout()
plt.savefig("edge_heatmap_all_targets2.png", dpi=200)
plt.show()


fig2, ax2 = plt.subplots(figsize=(12, 6))
im2 = ax2.imshow(node_feature_matrix, aspect="auto", cmap="Blues")
ax2.set_xticks(range(len(FEATURE_NAMES)))
ax2.set_xticklabels(FEATURE_NAMES, rotation=45, ha="right", fontsize=9)
ax2.set_yticks(range(len(TARGET_NAMES)))
ax2.set_yticklabels([f"{n} ({u})" for n, u in zip(TARGET_NAMES, TARGET_UNITS)], fontsize=9)
ax2.set_title("Node feature importance\n(GNNExplainer + GraphSAGE)")
plt.colorbar(im2, ax=ax2, label="importance")
plt.tight_layout()
plt.savefig("node_feature_heatmap2.png", dpi=200)
plt.show()


fig3, axes = plt.subplots(4,5,figsize=(20, 16))
axes = axes.flatten()


for idx, (target_name, unit) in enumerate(zip(TARGET_NAMES, TARGET_UNITS)):
    ax = axes[idx]
    edge_weights = aggregated_edge_masks[idx]
    threshold = np.quantile(edge_weights, 0.80)
    edge_weights_thr = np.where(edge_weights >= threshold, edge_weights, 0)
    ew = edge_weights_thr.copy()
    if ew.max() > 0:
        ew = (ew - ew.min()) / (ew.max() - ew.min() + 1e-8)
    node_importance = node_masks_all[idx].mean(axis=1)
    node_importance = (node_importance - node_importance.min()) / (node_importance.max() - node_importance.min() + 1e-8)

     nx.draw_networkx_edges(G, pos, ax=ax, edgelist=list(G.edges()),
                           width=[w * 5 + 0.2 for w in ew], edge_color=ew,
                           edge_cmap=plt.cm.Oranges, alpha=0.9)
    nx.draw_networkx_nodes(G, pos, ax=ax, node_color=node_importance,
                           cmap=plt.cm.Greens, node_size=220)
    nx.draw_networkx_labels(G, pos, ax=ax, font_size=7)
    ax.set_title(f"{target_name}\n({unit})", fontsize=9)
    ax.axis("off")


if len(TARGET_NAMES) < len(axes):
    for j in range(len(TARGET_NAMES), len(axes)):
        axes[j].set_visible(False)

fig3.suptitle("GraphSAGE + GNNExplainer", fontsize=14)
plt.tight_layout()
plt.savefig("all_targets_subplots2.png", dpi=200)
plt.show()
plt.show()
```
