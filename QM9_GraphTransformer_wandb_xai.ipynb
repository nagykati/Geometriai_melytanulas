```python
device = "cuda" if torch.cuda.is_available() else "cpu"
```

```python
dataset = QM9(root ="data/qm9")
```

```python
print(f"  Number of graphs: {len(dataset)}")
print(f"  Number of node features: {dataset.num_node_features}")
print(f"  Number of edge features: {dataset.num_edge_features}")
print(f"  Number of targets: {dataset.num_classes}")
print(f"  Number of features: {dataset.num_features}")

sample = dataset[0]
print(f"\nExample molecule:")
print(f"  Name       : {sample.name}")
print(f"  Atoms      : {sample.x.shape[0]}")
print(f"  x shape    : {sample.x.shape}   (atoms x features)")
print(f"  edge_index : {sample.edge_index.shape}")
print(f"  edge_attr  : {sample.edge_attr.shape}")
print(f"  pos        : {sample.pos.shape}")
print(f"  y          : {sample.y}")
```

```python
from torch.utils.data import random_split

num_graphs = len(dataset)
train_size = int(0.8 * num_graphs)
validation_size = int(0.1 * num_graphs)
test_size = num_graphs - train_size - validation_size

torch.manual_seed(42)

train_dataset, validation_dataset, test_dataset = random_split(
    dataset,
    [train_size, validation_size, test_size]
)

HIDDEN_DIM  = 64
```

```python
NUM_TARGETS = 19

TARGET_NAMES = [
    "mu", "alpha", "HOMO", "LUMO", "gap", "R2", "ZPVE",
    "U0", "U", "H", "G", "Cv",
    "U0_atom", "U_atom", "H_atom", "G_atom",
    "A", "B", "C"
]

TARGET_UNITS = [
    "D", "a0^3", "eV", "eV", "eV", "a0^2", "eV",
    "eV", "eV", "eV", "eV", "cal/mol/K",
    "eV", "eV", "eV", "eV",
    "GHz", "GHz", "GHz"
]
```

```python
all_train_y = torch.cat([data.y for data in train_dataset], dim=0)

target_mean = all_train_y.mean(dim=0)
target_std  = all_train_y.std(dim=0).clamp(min=1e-6)

print("Per-target statistics (train set):")
print(f"{'idx':>4} {'name':>10} {'mean':>12} {'std':>12} {'unit'}")
print("-" * 55)
for i, (name, unit) in enumerate(zip(TARGET_NAMES, TARGET_UNITS)):
    print(f"{i:>4} {name:>10} {target_mean[i].item():>12.4f} {target_std[i].item():>12.4f}  {unit}")
```

```python
def normalize_dataset(subset):
    normalized = []
    for data in subset:
        data.y = (data.y - target_mean) / target_std
        normalized.append(data)
    return normalized

train_dataset      = normalize_dataset(train_dataset)
validation_dataset = normalize_dataset(validation_dataset)
test_dataset       = normalize_dataset(test_dataset)
```

```python
batch_size = 64
train_loader = DataLoader(
    train_dataset,
    batch_size=batch_size,
    shuffle=True)

validation_loader = DataLoader(
    validation_dataset,
    batch_size=batch_size,
    shuffle=False)

test_loader = DataLoader(
    test_dataset,
    batch_size=batch_size,
    shuffle=False)

print(f"Train batches : {len(train_loader)}")
print(f"Validation   batches : {len(validation_loader)}")
print(f"Test  batches : {len(test_loader)}")
```

```python
class GraphTransformer(torch.nn.Module):
    def __init__(self):
        super().__init__()

        # Input embedding
        self.input_lin = Linear(dataset.num_node_features, HIDDEN_DIM)

        self.transf1 = TransformerConv(HIDDEN_DIM, HIDDEN_DIM, edge_dim = 4)
        self.bn1   = BatchNorm1d(HIDDEN_DIM)
        self.transf2 = TransformerConv(HIDDEN_DIM, HIDDEN_DIM, edge_dim = 4)
        self.bn2   = BatchNorm1d(HIDDEN_DIM)

        self.lin1 = Linear(HIDDEN_DIM, HIDDEN_DIM // 2)
        self.lin2 = Linear(HIDDEN_DIM // 2, NUM_TARGETS)

    def forward(self, x, edge_index, edge_attr, batch):
        x = self.input_lin(x).relu()

        x, attn1 = self.transf1(x, edge_index, edge_attr, return_attention_weights=True)
        x = self.bn1(x).relu()

        x, attn2 = self.transf2(x, edge_index, edge_attr, return_attention_weights=True)
        x = self.bn2(x).relu()

        x = global_mean_pool(x, batch)

        x = self.lin1(x).relu()
        x = self.lin2(x)
        return x

model = GraphTransformer().to(device)

print(model)
print(f"\nTrainable parameters: {sum(p.numel() for p in model.parameters() if p.requires_grad):,}")
```

```python
optimizer = torch.optim.Adam(model.parameters(), lr=0.001, weight_decay=1e-5)

scheduler = torch.optim.lr_scheduler.ReduceLROnPlateau(
    optimizer,
    mode='min',
    factor=0.75,
    patience=10,
    min_lr=1e-6
)

_mean = target_mean.to(device)
_std  = target_std.to(device)


def train(loader):
    model.train()
    total_loss = 0
    for batch in loader:
        batch = batch.to(device)
        optimizer.zero_grad()
        out  = model(batch.x, batch.edge_index, batch.edge_attr, batch.batch)
        loss = F.mse_loss(out, batch.y)
        loss.backward()
        torch.nn.utils.clip_grad_norm_(model.parameters(), 1.0)
        optimizer.step()
        total_loss += loss.item() * batch.num_graphs
    return total_loss / len(loader.dataset)


@torch.no_grad()
def evaluate(loader):
    model.eval()

    total_mae = torch.zeros(NUM_TARGETS)

    for batch in loader:
        batch = batch.to(device)
        out   = model(batch.x, batch.edge_index, batch.edge_attr, batch.batch)

        pred = out  * _std + _mean
        true = batch.y * _std + _mean

        total_mae += (pred - true).abs().sum(dim=0).cpu()

    mae_per_target = total_mae / len(loader.dataset)
    mean_mae       = mae_per_target.mean().item()
    return mae_per_target, mean_mae
```

```python
wandb.init(
    project="qm9-GraphTransformer",
    config={
        "model": "graphTransformer",
        "hidden_dim": 64,
        "num_layers": 2,
        "lr": 0.001,
        "batch_size": 64,
        "epochs": 100,
        "optimizer": "Adam",
        "num_node_features": dataset.num_node_features,
        "trainable_params": sum(p.numel() for p in model.parameters() if p.requires_grad),
        "normalization": "per_target_zscore",
        "pooling": "global_mean_pool",
    }
)

wandb.watch(model, log="all", log_freq=10)
```

```python
best_val_mae = float('inf')
best_test_mae_per_target = None

history = {'train_loss': [], 'val_mae': [], 'test_mae': [], 'lr': []}

for epoch in range(1, 101):
    train_loss              = train(train_loader)
    val_mae_per_t, val_mae  = evaluate(validation_loader)
    test_mae_per_t, test_mae = evaluate(test_loader)

    scheduler.step(val_mae)

    current_lr = optimizer.param_groups[0]['lr']

    history['train_loss'].append(train_loss)
    history['val_mae'].append(val_mae)
    history['test_mae'].append(test_mae)
    history['lr'].append(current_lr)

    log_dict = {
            "epoch": epoch,
            "train_loss": train_loss,
            "val_mae": val_mae,
            "test_mae": test_mae,
            "lr": current_lr

    }
    for i, name in enumerate(TARGET_NAMES):
        log_dict[f"val_mae/{name}"]  = val_mae_per_t[i].item()
        log_dict[f"test_mae/{name}"] = test_mae_per_t[i].item()

    wandb.log(log_dict)

    if val_mae < best_val_mae:
        best_val_mae             = val_mae
        best_test_mae_per_target = test_mae_per_t.clone()
        torch.save(model.state_dict(), 'best_model.pt')
        wandb.save('best_model.pt')

    if epoch % 10 == 0:
        print(f"Epoch {epoch:>3} | Loss: {train_loss:.4f} | "
              f"Val MAE: {val_mae:.4f} | Test MAE: {test_mae:.4f} | LR: {current_lr:.2e}")

print(f"\nBest mean Val MAE  : {best_val_mae:.4f}")
```

```python
import networkx as nx
from collections import defaultdict

model.load_state_dict(torch.load('best_model.pt'))
model.eval()

sample = test_dataset[0].to(device)

with torch.no_grad():
    x = model.input_lin(sample.x).relu()
    x, (edge_idx1, attn1) = model.transf1(x, sample.edge_index, sample.edge_attr, return_attention_weights=True)
    x = model.bn1(x).relu()
    x, (edge_idx2, attn2) = model.transf2(x, sample.edge_index, sample.edge_attr, return_attention_weights=True)

attn1_mean = attn1.mean(dim=1).cpu().numpy()
attn2_mean = attn2.mean(dim=1).cpu().numpy()
```

```python
def aggregate_undirected(edge_index, attn_weights):
    edge_groups = defaultdict(list)
    ei = edge_index.cpu().numpy()
    for i in range(ei.shape[1]):
        u, v = int(ei[0, i]), int(ei[1, i])
        edge_groups[tuple(sorted((u, v)))].append(attn_weights[i])
    undir_edges = list(edge_groups.keys())
    undir_weights = np.array([np.mean(v) for v in edge_groups.values()])
    return undir_edges, undir_weights

undir_edges1, uw1 = aggregate_undirected(edge_idx1, attn1_mean)
undir_edges2, uw2 = aggregate_undirected(edge_idx2, attn2_mean)

G = nx.Graph()
G.add_edges_from(undir_edges1)
pos = nx.spring_layout(G, seed=42)
```

```python
def draw_attention(ax, G, pos, undir_edges, undir_weights, title):
    ew = undir_weights.copy()
    if ew.max() > 0:
        ew = (ew - ew.min()) / (ew.max() - ew.min() + 1e-8)
    nx.draw_networkx_edges(G, pos, ax=ax, edgelist=undir_edges,
                           width=[w * 6 + 0.3 for w in ew],
                           edge_color=ew, edge_cmap=plt.cm.Reds,
                           edge_vmin=0, edge_vmax=1, alpha=0.9)
    nx.draw_networkx_nodes(G, pos, ax=ax, node_color="lightblue", node_size=300)
    nx.draw_networkx_labels(G, pos, ax=ax, font_size=8)
    ax.set_title(title, fontsize=11)
    ax.axis("off")

fig, (ax1, ax2) = plt.subplots(1, 2, figsize=(14, 6))
draw_attention(ax1, G, pos, undir_edges1, uw1, "1. réteg attention")
draw_attention(ax2, G, pos, undir_edges2, uw2, "2. réteg attention")
fig.suptitle("GraphTransformer — Attention weights (best_model.pt)", fontsize=13)
plt.tight_layout()
plt.savefig("attention_layers.png", dpi=200)
plt.show()
```
